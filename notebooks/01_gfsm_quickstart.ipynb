# GFSM v1 Quickstart

This notebook reads a local GFSM GeoTIFF tile, displays metadata, plots the classes, and summarizes class counts. Set `tile_path` to a tile downloaded from the Zenodo archive.

```python
from pathlib import Path

import matplotlib.colors as colors
import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import numpy as np
import rasterio

tile_path = Path('path/to/GFSM_tile.tif')
```

```python
with rasterio.open(tile_path) as src:
    data = src.read(1, masked=False)
    metadata = {
        'crs': src.crs,
        'resolution': src.res,
        'bounds': src.bounds,
        'shape': (src.height, src.width),
        'nodata': src.nodata,
    }

metadata
```

```python
values, counts = np.unique(data, return_counts=True)
dict(zip(values.tolist(), counts.tolist()))
```

```python
palette = ['#2c7bb6', '#abd9e9', '#ffffbf', '#fdae61', '#d7191c']
class_names = {1: 'Very Low', 2: 'Low', 3: 'Moderate', 4: 'High', 5: 'Very High'}
masked = np.ma.masked_where(data == 0, data)
cmap = colors.ListedColormap(palette)
norm = colors.BoundaryNorm([1, 2, 3, 4, 5, 6], cmap.N)

fig, ax = plt.subplots(figsize=(8, 6))
ax.imshow(masked, cmap=cmap, norm=norm, interpolation='nearest')
ax.set_axis_off()
ax.set_title('GFSM v1 Flood Susceptibility')
ax.legend(
    handles=[mpatches.Patch(color=palette[v - 1], label=f'{v} - {n}') for v, n in class_names.items()],
    loc='lower left'
)
plt.show()
```
